```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
netflix=pd.read_csv('../data/netflix_titles.csv')
```

```python
netflix.director.fillna(value='unknown',inplace=True)
netflix.director
```

```python
netflix.cast.fillna(value='unknown',inplace=True)
netflix.cast
```

```python
netflix.country.fillna(value='unknown',inplace=True)
netflix.country
```

```python
netflix.date_added.fillna(value='unknown',inplace=True)
netflix.date_added
```

```python
netflix.isnull().sum()
```

```python
netflix.dropna(inplace=True)
```

```python
netflix.isnull().sum()
```

```python
tcount=netflix["type"].value_counts()
```

```python
tcount
```

```python
plt.figure()
tcount.plot(kind='bar')
plt.title("Content count by type")
plt.xlabel("Type")
plt.ylabel("Count")
plt.savefig('content_count.png')
plt.show()
```

```python
# from above we can interpret that Movies are more frequent than Tv shows in the dataset
```

```python
netflix['release_year']=pd.to_numeric(netflix['release_year'],errors='coerce')
ycount=netflix['release_year'].value_counts().sort_index()

plt.figure()
ycount.plot()
plt.title('Content Growth')
plt.xlabel('Year')
plt.ylabel('Number of titles')
plt.savefig('growth_count.png')
plt.show()
```

```python
# from above we interpret the content growth significantly increased after around 2015, showing rapid expansion
```

```python
genres=netflix['listed_in'].dropna().str.split(", ")
all_genres=genres.explode()
tgenres=all_genres.value_counts().head(10)
tgenres
```

```python
plt.figure()
tgenres.plot(kind='bar')
plt.title('Top 10 genres')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.savefig('top_genres.png')
plt.show()
```

```python
# the most common genres are 'International Movies','Dramas','Comedies'
```

```python
netflix['runtime']=netflix['duration'].str.extract("(\d+)").astype(float)
plt.figure()
sns.histplot(netflix['runtime'].dropna(),bins=30,kde=True)
plt.savefig('runtime.png')
plt.show()
```

```python
# the runtime distribution shows most titles fall between 80-130 mins
```

```python
tyear=netflix['release_year'].value_counts().head(10)
plt.figure()
tyear.plot(kind='bar')
plt.xlabel('Year')
plt.ylabel('Count')
plt.savefig('top.png')
plt.show()
```

```python
# the most actve release years are: 2018, 2017 & 2019 had same releases
```
